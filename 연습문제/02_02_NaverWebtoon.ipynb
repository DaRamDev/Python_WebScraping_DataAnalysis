```python
import requests
from bs4 import BeautifulSoup
import os

def download_one_episode(title, no, url):
    headers = {
        'Referer': url,
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    
    save_dir = os.path.join('img', title, str(no))
    os.makedirs(save_dir, exist_ok=True)
    
    res = requests.get(url, headers=headers)
    if not res.ok:
        print(f"페이지 요청 실패 (Error: {res.status_code})")
        return

    soup = BeautifulSoup(res.text, 'html.parser')
    img_tags = soup.select("img[src*='IMAG01']")
    
    if not img_tags:
        img_tags = soup.select("div.wt_viewer img")

    print(f"총 {len(img_tags)}개의 이미지를 다운로드합니다.")

    for img_tag in img_tags:
        img_url = img_tag.get('src')
        if not img_url:
            continue
            
        file_name = os.path.basename(img_url)
        file_path = os.path.join(save_dir, file_name)
        
        img_res = requests.get(img_url, headers=headers)
        if img_res.ok:
            with open(file_path, 'wb') as file:
                file.write(img_res.content)
            print(f"저장 완료: {file_path} ({len(img_res.content):,} bytes)")

download_one_episode('일렉시드', 341, 'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')
```
