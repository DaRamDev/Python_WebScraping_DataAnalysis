```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.daum.net/economy'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=headers)
res.encoding = 'utf-8'

print(url)
print(type(res))
print(res.status_code)

soup = BeautifulSoup(res.text, 'html.parser')
li_tags = soup.select('ul.list_newsheadline2 li')
print(type(li_tags), len(li_tags))

for li_tag in li_tags:
    a_tag = li_tag.find('a')
    if not a_tag:
        continue
    link = a_tag['href']
    
    strong_tag = li_tag.select_one('strong.tit_txt')
    if strong_tag:
        title = strong_tag.text.strip()
        print(link)
        print(title)
```

```python
import requests
from bs4 import BeautifulSoup

section_dict = {
    '기후/환경': 'climate', '사회': 'society', '경제': 'economy', '정치': 'politics',
    '국제': 'world', '문화': 'culture', '생활': 'life', 'IT/과학': 'tech', '인물': 'people'
}

def print_news(section_name):
    section_path = section_dict.get(section_name)
    if not section_path:
        print("존재하지 않는 섹션입니다.")
        return

    url = f'https://news.daum.net/{section_path}'
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=headers)
    res.encoding = 'utf-8'

    if res.ok:
        print(f"======> {url} {section_name} 뉴스 <======")
        soup = BeautifulSoup(res.text, 'html.parser')
        li_tags = soup.select('ul.list_newsheadline2 li')

        for li_tag in li_tags:
            a_tag = li_tag.find('a')
            if not a_tag:
                continue
            link = a_tag['href']
            strong_tag = li_tag.select_one('strong.tit_txt')
            if strong_tag:
                title = strong_tag.text.strip()
                print(link)
                print(title)
    else:
        print(f"Error Code = {res.status_code}")


print_news('경제')
print_news('사회')
```
